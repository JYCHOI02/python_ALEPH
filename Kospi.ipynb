```python
from flask import Flask, render_template_string
import yfinance as yf
from datetime import datetime
import pytz

app = Flask(__name__)

HTML_TEMPLATE = """
<!DOCTYPE html>
<html lang="ko">
<head>
  <meta charset="UTF-8">
  <title>KOSPI 지수 고정 화면</title>
  <style>
    body { font-family: sans-serif; background-color: #f4f6f9; display: flex; justify-content: center; align-items: center; height: 100vh; margin: 0; }
    .card { background: white; padding: 30px; border-radius: 12px; box-shadow: 0 4px 15px rgba(0,0,0,0.1); width: 320px; }
    .title { font-size: 14px; color: #666; font-weight: bold; margin-bottom: 8px; }
    .value-container { display: flex; align-items: baseline; margin-bottom: 20px; border-bottom: 2px solid #eee; padding-bottom: 15px; }
    .value { font-size: 36px; font-weight: bold; color: #111; margin-right: 8px; }
    .unit { font-size: 18px; color: #555; font-weight: 500; }
    .info-list { list-style: none; padding: 0; margin: 0; font-size: 13px; color: #444; line-height: 1.8; }
    .info-list li { display: flex; justify-content: space-between; }
    .label { color: #888; }
  </style>
</head>
<body>
  <div class="card">
    <div class="title">코스피 지수 (KOSPI)</div>
    <div class="value-container">
      <span class="value">{{ price }}</span>
      <span class="unit">pt</span>
    </div>
    <ul class="info-list">
      <li><span class="label">출처</span><span>한국거래소 (KRX)</span></li>
      <li><span class="label">데이터 시각</span><span>{{ data_time }}</span></li>
      <li><span class="label">조회 시각</span><span>{{ fetch_time }}</span></li>
      <li><span class="label">기준 시간대</span><span>KST (UTC+09:00)</span></li>
    </ul>
  </div>
</body>
</html>
"""

@app.route('/')
def home():
    # KOSPI 데이터 가져오기
    ticker = yf.Ticker("^KS11")
    info = ticker.fast_info
    
    current_price = f"{info.last_price:,.2f}"
    
    # 시간 설정 (KST)
    tz_kst = pytz.timezone('Asia/Seoul')
    now = datetime.now(tz_kst).strftime('%Y-%m-%d %H:%M:%S')
    
    return render_template_string(HTML_TEMPLATE, 
                                  price=current_price, 
                                  data_time=now, 
                                  fetch_time=now)

if __name__ == '__main__':
    app.run(port=5000, debug=True)
```
